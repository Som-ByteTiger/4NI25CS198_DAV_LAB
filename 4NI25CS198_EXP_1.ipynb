```python
import numpy as np
import pandas as pd
marks=np.array([78,85,92,67,88])
print("Marks :",marks)
print("Type :",type(marks))
print("Number of dimensions : ",marks.ndim)
print("Add 5 : ",marks+5)
print("Multiply by 2 : ",marks*2)
print("Divide by 2 : ",marks/2)
print("Square : ",marks**2)
print("Sum : ",marks.sum())
print("mean : ",marks.mean())
print("Maximum : ",marks.max())
print("Minimum : ",marks.min())
print("First marks : ",marks[0])
print("Second marks : ",marks[1])
print("Last marks : ",marks[-1])
s_m=np.array([
    [78,86,89],
    [99,78,67],
    [98,98,76],
])
print("Shape",s_m.shape)
print("Column mean : ",s_m.mean(axis=0))
print("Row mean : ",s_m.mean(axis=1))
```

```python
import numpy as np
import pandas as pd
data={
    "Name":["asha","Bharat","chetan","Divya","Esha"],
    "Department":["CSE","ISE","CSE","AIML","ISE"],
    "Marks":[72,91,84,95,68],
    "Attendance":[82,94,88,96,75]
}
students=pd.DataFrame(data)
print(students[['Name', 'Marks']])
print(students[students['Marks'] > 80])
print(students[students['Attendance'] > 80])
print(students.head(3))
avg_marks = students['Marks'].mean()
max_marks = students['Marks'].max()
print(avg_marks)
print(max_marks)
cs_students = students[students['Department'] == 'CSE']
print(cs_students)
result = students[(students['Attendance'] > 85) & (students['Marks'] > 80)]
```
